## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Data Wrangling

# Create a copy of the original dataset
df_clean = df.copy()

# Standardize Gender values
df_clean['Gender'] = df_clean['Gender'].astype(str).str.strip().str.lower()

df_clean['Gender'] = df_clean['Gender'].replace({
    'm': 'Male',
    'male': 'Male',
    'man': 'Male',
    'female': 'Female',
    'f': 'Female',
    'woman': 'Female'
})

# Handle missing values in important categorical columns
for col in ['state', 'self_employed', 'work_interfere']:
    df_clean[col] = df_clean[col].fillna('Unknown')

# Remove unrealistic Age values
df_clean = df_clean[
    (df_clean['Age'] >= 18) &
    (df_clean['Age'] <= 100)
]

# Remove duplicate rows
df_clean = df_clean.drop_duplicates()

# Display cleaned dataset information
print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", df_clean.shape)
print("\nRemaining missing values:")
print(df_clean.isnull().sum())
````

### What all manipulations have you done and insights you found?

The dataset was cleaned and prepared for further exploratory analysis. Gender values were standardized to reduce inconsistencies caused by different labels and spellings. Missing values in important variables such as state, self_employed, and work_interfere were handled by assigning them as "Unknown". Unrealistic age values below 18 or above 100 were removed, and duplicate records were checked and removed.

The initial analysis shows that the dataset contains substantial missing information in the comments variable, while most important mental-health and workplace variables are comparatively complete. The cleaned dataset is now more consistent and suitable for analysing factors associated with mental-health treatment, workplace support, and employee attitudes.

#### Chart - 14 - Correlation Heatmap

````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("survey.csv")

# Select numeric columns
numeric_df = df.select_dtypes(include="number")

# Calculate correlation
corr = numeric_df.corr()

# Create heatmap using matplotlib
fig, ax = plt.subplots(figsize=(8, 6))
heatmap = ax.imshow(corr, aspect="auto")

ax.set_xticks(range(len(corr.columns)))
ax.set_yticks(range(len(corr.columns)))
ax.set_xticklabels(corr.columns, rotation=45, ha="right")
ax.set_yticklabels(corr.columns)

# Display correlation values
for i in range(len(corr.columns)):
    for j in range(len(corr.columns)):
        ax.text(j, i, f"{corr.iloc[i, j]:.2f}",
                ha="center", va="center")

plt.title("Correlation Heatmap of Numerical Variables")
plt.colorbar(heatmap)
plt.tight_layout()
plt.show()
````

##### 1. Why did you pick the specific chart?

A correlation heatmap was selected because it provides a clear visual representation of the strength and direction of relationships between numerical variables, making patterns and correlations easier to identify.

##### 2. What is/are the insight(s) found from the chart?

The heatmap shows the correlations among the numerical variables in the dataset. Most variables do not show strong linear relationships with each other, suggesting that the numerical characteristics operate relatively independently and no major multicollinearity is evident.
